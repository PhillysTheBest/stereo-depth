```python
%%writefile hello_cuda.cu
#include <stdio.h>

__global__ void hello() {
    printf("Hello from block: %d, thread: %d\n", blockIdx.x, threadIdx.x);
}

int main() {
    hello<<<2, 2>>>();
    cudaDeviceSynchronize();
    return 0;
}
```

```python
!nvcc hello_cuda.cu -o hello_cuda
!./hello_cuda
```
